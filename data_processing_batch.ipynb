```pyspark
from pyspark.sql import SparkSession 
from pyspark.sql.functions import col, to_timestamp, current_timestamp

# Czytanie surowego pliku
input_batch = spark.read.json("s3://emr-notebook-as-2137/raw/")

# Przetworzenie czasu pobrania na timestamp oraz dodanie czasu przetworzenia
processed_batch = input_batch.withColumn("event_time", to_timestamp(col("timestamp"))).withColumn("processing_time", current_timestamp()).drop("timestamp")

# Nadpisywanie przetworzonych danych w formacie parquet
processed_batch.write \
    .format("parquet") \
    .mode("overwrite") \
    .save("s3://emr-notebook-as-2137/curated_batch/")

print("Przetwarzanie zakonczone")
```
