```python
import requests
import boto3
import time
import json
from datetime import datetime

API_URL = "https://e6uw49pbah.execute-api.us-east-1.amazonaws.com/dev/weather/latest"
HEADERS = {
    "Authorization": "Bearer STUDENT_TOKEN_2026"
}
PARAMS = {
    "station_id": "GDN_01"
}

s3_bucket = "emr-notebook-as-2137"
s3_client = boto3.client('s3')

while True:
    response = requests.get(API_URL, headers=HEADERS, params=PARAMS)
    # Poprawnie wyslane zapytanie i odebranie odpowiedzi jesli kod 200
    if response.status_code == 200:
        data = response.json()
        current_time = datetime.now()
        timestamp_str = current_time.strftime("%Y%m%d_%H%M%S")
        # Nazwa pliku z aktualnym czasem
        file_name = f"raw/weather_{PARAMS['station_id']}_{timestamp_str}.json"
            
        # Zapisanie surowego pliku json w buckecie
        s3_client.put_object(
            Bucket=s3_bucket,
            Key=file_name,
            Body=json.dumps(data)
        )
        
        print(f"Saved: {file_name}")
    else:
        print(f"Error: {response.status_code} - {response.text}")
        
    time.sleep(5)
```
